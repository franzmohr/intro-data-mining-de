## Pakete installieren

Der Befehl `%pip install` installiert die Pakete in genau die Python-Umgebung, in der dieser Jupyter-Kernel läuft. Alternativ lässt sich derselbe Befehl ohne `%` in einem Terminal ausführen: `python -m pip install ...`.

```python
%pip install geopandas kagglehub matplotlib mlxtend networkx numpy pandas pyproj scikit-learn scipy seaborn torch jupyterlab
```

Die Installation dauert beim ersten Mal einige Minuten, vor allem wegen `torch`. Sind die Pakete schon vorhanden, meldet pip für jedes Paket `Requirement already satisfied`. Damit neu installierte Pakete gefunden werden, sollte der Kernel danach neu gestartet werden (*Kernel → Restart Kernel*).

## Installation prüfen

Die folgende Zelle lädt alle Pakete und gibt ihre Versionen aus. Erscheint ein `ModuleNotFoundError`, ist das betreffende Paket nicht installiert.

```python
from importlib.metadata import version

import geopandas, kagglehub, matplotlib, mlxtend, networkx, numpy, pandas, pyproj, sklearn, scipy, seaborn, torch

for paket in ["geopandas", "kagglehub", "matplotlib", "mlxtend", "networkx", "numpy",
              "pandas", "pyproj", "scikit-learn", "scipy", "seaborn", "torch"]:
    print(f"{paket:15} {version(paket)}")
```
