## Datenvisualisierung mit `matplotlib`

`matplotlib` (https://matplotlib.org/) ist ein äußerst beliebtes Paket für die Erstellung von Grafiken mit Python. Die Grafikfunktionalität des Pakets kann auf folgende Weise geladen werden:

```python
# Lade das Modul pyplot aus dem Paket matplotlib und mach es über die
# Abkürzung plt verfügbar
import matplotlib.pyplot as plt
import pandas
```

```python
# Lade einen Beispieldatensatz
pwt = pandas.read_excel("pwt100.xlsx",
                        sheet_name = "Data")

# Filter nach AT
pwt_at = pwt[pwt["countrycode"] == "AUT"]

# Wähle einzelne Spalten aus
pwt_at = pwt_at[["year", "countrycode", "rgdpe", "pop", "emp", "hc"]]

# Zeige Ergebnis an
pwt_at.head()
```

`matplotlib`-Grafiken folgen einer Art Baukastenprinzip, wobei jeder Baukasten mit dem Befehl `plt.figure()` beginnt. Damit wird ein leeres Grafik-Objekt erstellt. Noch genauer: Der Befehl besagt lediglich, *dass* eine Grafik erstellt werden soll.

```python
fig = plt.figure()
```

Mit dem folgenden Befehl wird schließlich ein leeres Grafik-Objekt innerhalb von `fig` erstellt:

```python
ax = fig.add_subplot()
```

*NB: Mit dem Befehl `fig.add_subplot(2, 3, 1)` würden 2 x 3 leere Grafikobjekte erzeugt werden, die separat befüllt werden können. Das ist allerdings ein Thema für einen Folgekurs.*

Mit dem Befehl `ax.plot` kann das leere Grafikobjekt mit Inhalt befüllt werden.

```python
ax.plot(pwt_at.loc[:, "year"], pwt_at.loc[:, "rgdpe"])
```

Aus technischen Gründen müssen die drei beschriebenen Zeilen in Jupyter in einer Notebook-Zelle ausgeführt werden:

```python
fig = plt.figure()
ax = fig.add_subplot()
ax.plot(pwt_at.loc[:, "year"], pwt_at.loc[:, "rgdpe"])
```
